```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
df = pd.read_csv("../results/batch_results.csv")
df.head()
```

```python
len(df)
```

```python
f, axes = plt.subplots(2, 1, figsize=(16, 8))
sb.boxplot(data=df["x"], ax=axes[0], orient="h")
sb.boxplot(data=df["theta"], ax=axes[1], orient="h")
```

```python
successful_runs = df.loc[df["success"]==True]
successful_runs.head()
```

```python
failed_runs = df.loc[df["success"]==False]
failed_runs.head()
```

```python
len(failed_runs)
```

```python
f = plt.figure(figsize=(16, 8))
# plt.scatter(successful_runs["x"], successful_runs["theta"], color="green")
plt.scatter(failed_runs["x"], failed_runs["theta"], color="red")
plt.ylim((-180, 180))
```
